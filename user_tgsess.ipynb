```python
#@title <center>Telegram UserSession</center>

#@markdown ---
#@markdown #### API Credentials

API_KEY = None #@param {type:"integer"}
API_HASH = "" #@param {type:"string"}

#@markdown ---
#@markdown #### Library

Mode = "wzgram" #@param ["wzgram", "pyrogram", "pyrofork", "pyrotgfork", "telethon"] {allow-input: false, type:"string"}

#@markdown ---
#@markdown #### Version

Wzgram_Version = "none" #@param ["none", "latest"] {allow-input: true, type:"string"}
Pyrogram_Version = "none" #@param ["none", "2.0.66", "2.0.77", "2.0.106", "latest"] {allow-input: true, type:"string"}
Pyrofork_Version = "none" #@param ["none", "2.2.10", "2.3.68", "latest"] {allow-input: true, type:"string"}
Pyrotgfork_Version = "none" #@param ["none", "2.2.17", "latest"] {allow-input: true, type:"string"}
Telethon_Version = "none" #@param ["none", "latest"] {allow-input: true, type:"string"}

if Mode == "wzgram":
    v = Wzgram_Version if Wzgram_Version != "none" else "latest"
    !pip install wzgram
    header = f"#WZGRAM_SESSION_{v}"
elif Mode == "pyrogram":
    v = Pyrogram_Version if Pyrogram_Version != "none" else "latest"
    pg = f"pyrogram=={v}"
    if v == "latest":
        pg = "pyrogram"
    !pip3 install tgcrypto $pg
    header = f"#PYROGRAM_SESSION_{v}"
elif Mode == "pyrofork":
    v = Pyrofork_Version if Pyrofork_Version != "none" else "latest"
    pf = f"pyrofork=={v}"
    if v == "latest":
        pf = "pyrofork"
    !pip3 install tgcrypto $pf
    header = f"#PYROFORK_SESSION_{v}"
elif Mode == "pyrotgfork":
    v = Pyrotgfork_Version if Pyrotgfork_Version != "none" else "latest"
    ptg = f"pyrotgfork=={v}"
    if v == "latest":
        ptg = "pyrotgfork"
    !pip3 install tgcrypto $ptg
    header = f"#PYROTGFORK_SESSION_{v}"
elif Mode == "telethon":
    v = Telethon_Version if Telethon_Version != "none" else "latest"
    tt = f"telethon=={v}"
    if v == "latest":
        tt = "telethon"
    !pip3 install cryptg $tt
    header = f"#TELETHON_SESSION_{v}"

from IPython.display import clear_output
clear_output()
print(f"Successfully Installed {header}")

if Mode == "telethon":
    from telethon import TelegramClient
    from telethon.sessions import StringSession

    async with TelegramClient(StringSession(), API_KEY, API_HASH) as user:
        sess = user.session.save()
        print("\nGenerating ...")
        donestr = "sent to Saved Messages !!"
        try:
            await user.send_message('me',
                f"#WZMLX {header}\n\n`{sess}`",
            )
        except:
            donestr = "Successfully Printed !!"
            print(f"#WZMLX {header}")
            print(sess)
        print(f"Done !!, String Session has been {donestr}")
else:
    from pyrogram import Client
    from pyrogram.errors import UserIsBot

    async with Client(name="WZ-User", api_id=API_KEY, api_hash=API_HASH, in_memory=True) as app:
        sess = await app.export_session_string()
        print("\nGenerating ...")
        donestr = "sent to Saved Messages !!"
        try:
            await app.send_message("me",
                 f"#WZMLX {header}\n\n<code>{sess}</code>",
            )
        except UserIsBot:
            donestr = "Successfully Printed !!"
            print(f"#WZMLX {header}")
            print(sess)
        print(f"Done !!, String Session has been {donestr}")
```
